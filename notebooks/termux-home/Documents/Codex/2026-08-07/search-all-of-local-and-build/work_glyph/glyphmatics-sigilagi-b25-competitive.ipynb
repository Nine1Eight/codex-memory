```python
if not _IS_SCORED_RERUN:
    print('[PACKAGE] baseline execution skipped; no scored output generated')
else:
    # Run the baseline through the official ARC competition harness.
    import json
    import os
    import shutil
    import subprocess
    import sys
    import time
    import urllib.request
    from pathlib import Path

    _existing_output = Path('/kaggle/working/submission.parquet')
    if _existing_output.exists():
        _existing_output.unlink()
    _work = Path('/kaggle/working/ARC-AGI-3-Agents')
    _input = Path('/kaggle/input/competitions/arc-prize-2026-arc-agi-3/ARC-AGI-3-Agents')
    if not _input.exists():
        raise FileNotFoundError(_input)

    _deadline = time.time() + 600
    while time.time() < _deadline:
        try:
            with urllib.request.urlopen('http://gateway:8001/api/games', timeout=5) as _response:
                if 200 <= _response.status < 300:
                    break
        except Exception:
            time.sleep(5)
    else:
        raise TimeoutError('ARC gateway unavailable')

    if _work.exists():
        shutil.rmtree(_work)
    shutil.copytree(_input, _work)
    _asset_root = Path('/kaggle/working/b25_baseline_assets')
    _template_dir = _work / 'agents/templates'
    _template_dir.mkdir(parents=True, exist_ok=True)
    for _name in ('my_agent.py', 'factory_exact_routes.json', 'public_exact_routes_from_replays.json'):
        shutil.copy2(_asset_root / 'agents/templates' / _name, _template_dir / _name)

    (_work / 'agents/__init__.py').write_text(
        'from typing import Type\n'
        'from dotenv import load_dotenv\n'
        'from .agent import Agent, Playback\n'
        'from .swarm import Swarm\n'
        'from .templates.random_agent import Random\n'
        'from .templates.my_agent import MyAgent\n'
        'load_dotenv()\n'
        'AVAILABLE_AGENTS: dict[str, Type[Agent]] = {"random": Random, "myagent": MyAgent}\n',
        encoding='utf-8',
    )
    (_work / '.env').write_text(
        'SCHEME=http\nHOST=gateway\nPORT=8001\nARC_API_KEY=test-key-123\n'
        'ARC_BASE_URL=http://gateway:8001/\nOPERATION_MODE=online\nENVIRONMENTS_DIR=\n'
        'RECORDINGS_DIR=/kaggle/working/server_recording\n',
        encoding='utf-8',
    )

    _env = os.environ.copy()
    _env.update({
        'ARC3_MAX_ACTIONS': '420',
        'P6_REPLAY_PREFIX': '1000000',
        'P6_USE_SHARED_MOVE_LIBRARY': '0',
        'ORPP_USE_BFS': '1',
        'MPLBACKEND': 'agg',
    })
    _run = subprocess.run(
        [sys.executable, 'main.py', '--agent', 'myagent'],
        cwd=_work,
        env=_env,
        text=True,
        capture_output=True,
        timeout=32400,
    )
    (_work / 'official_stdout.txt').write_text(_run.stdout or '', encoding='utf-8', errors='replace')
    (_work / 'official_stderr.txt').write_text(_run.stderr or '', encoding='utf-8', errors='replace')
    _text = (_run.stdout or '') + '\n' + (_run.stderr or '')
    if (_work / 'logs.log').exists():
        _text += '\n' + (_work / 'logs.log').read_text(encoding='utf-8', errors='replace')

    _marker = '--- FINAL SCORECARD REPORT ---'
    _pos = _text.rfind(_marker)
    _scorecard = None
    if _pos >= 0:
        _start = _text.find('{', _pos)
        if _start >= 0:
            try:
                _scorecard, _ = json.JSONDecoder().raw_decode(_text[_start:])
            except ValueError:
                _scorecard = None
    if _run.returncode != 0:
        raise RuntimeError('Official baseline rerun failed\n' + (_run.stderr or '')[-8000:])
    if not isinstance(_scorecard, dict):
        raise RuntimeError('Official harness produced no parseable scorecard')
    _envs = _scorecard.get('environments') or _scorecard.get('games') or []
    if len(_envs) != 25:
        raise RuntimeError(f'Expected 25 official game scorecard rows, got {len(_envs)}')

    import pandas as pd
    _rows = []
    for _game in _envs:
        _game_id = _game.get('id') or _game.get('game_id')
        if not _game_id:
            raise RuntimeError('Scorecard row has no game ID')
        _rows.append({
            'row_id': str(_game_id),
            'game_id': str(_game_id),
            'end_of_game': bool(_game.get('completed', _game.get('end_of_game', False))),
            'score': float(_game.get('score', _game.get('total_score', 0.0))),
        })
    if len({row['game_id'] for row in _rows}) != 25:
        raise RuntimeError('Official scorecard game IDs are not unique')
    pd.DataFrame(_rows, columns=['row_id', 'game_id', 'end_of_game', 'score']).to_parquet(
        '/kaggle/working/submission.parquet', index=False
    )
    print('[SCORED] 79.6404 baseline official rerun complete; rows=', len(_rows))
```
