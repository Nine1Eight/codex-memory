```python
import numpy as np
import json
import os
from scipy import ndimage
from typing import List, Dict, Any, Tuple

# =============================================================================
# TESSA5 KERNEL: RE-SOLVE / ARC-AGI-2 PROGRESS UNIT
# =============================================================================

class DSL:
    """Core Structural Primitives for the Valkyr Engine."""
    @staticmethod
    def rot90(x): return np.rot90(x, k=1)
    @staticmethod
    def rot180(x): return np.rot90(x, k=2)
    @staticmethod
    def flip_v(x): return np.flipud(x)
    @staticmethod
    def flip_h(x): return np.fliplr(x)
    @staticmethod
    def transpose(x): return x.T
    @staticmethod
    def fill_holes(x): 
        mask = (x > 0).astype(int)
        filled = ndimage.binary_fill_holes(mask).astype(int)
        return np.where(filled > mask, np.max(x) if np.max(x)>0 else 1, x)
    @staticmethod
    def crop_content(x):
        coords = np.argwhere(x > 0)
        if coords.size == 0: return x
        x_min, y_min = coords.min(axis=0)
        x_max, y_max = coords.max(axis=0)
        return x[x_min:x_max+1, y_min:y_max+1]

class Tessa5Kernel:
    def __init__(self):
        self.primitives = [
            DSL.rot90, DSL.rot180, DSL.flip_v, DSL.flip_h, 
            DSL.transpose, DSL.fill_holes, DSL.crop_content
        ]

    def extract_objects(self, grid):
        """L2-S: Topological Decomposition."""
        objects = []
        for color in range(1, 10):
            mask = (grid == color).astype(int)
            labels, num = ndimage.label(mask)
            for i in range(1, num + 1):
                objects.append({'color': color, 'mask': (labels == i)})
        return objects

    def synthesize(self, train_pairs: List[Dict]) -> Tuple[Any, float]:
        """Geodesic Mapping: Brute-force Search over Depth-1 and Depth-2 programs."""
        # Check Depth-1
        for op in self.primitives:
            match_count = 0
            for pair in train_pairs:
                inp, out = np.array(pair['input']), np.array(pair['output'])
                try:
                    res = op(inp)
                    if res.shape == out.shape and np.array_equal(res, out):
                        match_count += 1
                except: continue
            if match_count == len(train_pairs):
                return (lambda x: op(x)), 1.0

        # Check Depth-2 (Compositional)
        for op1 in self.primitives:
            for op2 in self.primitives:
                match_count = 0
                for pair in train_pairs:
                    inp, out = np.array(pair['input']), np.array(pair['output'])
                    try:
                        res = op2(op1(inp))
                        if res.shape == out.shape and np.array_equal(res, out):
                            match_count += 1
                    except: continue
                if match_count == len(train_pairs):
                    return (lambda x: op2(op1(x))), 1.0
        
        return (lambda x: x), 0.0 # Fallback: Identity

    def robust_guess(self, input_grid):
        """Attempt 2: Pattern completion fallback (Symmetry check)."""
        # Common ARC motif: Vertical or Horizontal symmetry completion
        h, w = input_grid.shape
        if h == w: return DSL.rot180(input_grid)
        return DSL.flip_h(input_grid)

def deploy():
    # 1. Environment Detection (ARC-AGI-2 2026 standard)
    DATA_ROOT = '/kaggle/input/arc-prize-2026-arc-agi-2/'
    INPUT_FILE = os.path.join(DATA_ROOT, 'arc-agi_test_challenges.json')
    
    # Fallback logic for various competition mount points
    if not os.path.exists(INPUT_FILE):
        INPUT_FILE = '/kaggle/input/competitions/arc-prize-2026-arc-agi-2/arc-agi_test_challenges.json'

    if not os.path.exists(INPUT_FILE):
        print("LOGIC DECOHERENCE: Challenges not found."); return

    with open(INPUT_FILE, 'r') as f:
        data = json.load(f)

    # 2. Reasoning Loop
    kernel = Tessa5Kernel()
    submission = {}

    for task_id, task_data in data.items():
        train = task_data.get('train', [])
        test = task_data.get('test', [])
        
        # Phase 1: Inductive Learning
        rule, χ_score = kernel.synthesize(train)
        
        # Phase 2: Deductive Inference
        task_predictions = []
        for instance in test:
            inp = np.array(instance['input'])
            
            # Attempt 1: The Logical Invariant
            ans1 = rule(inp)
            
            # Attempt 2: The Wyrd Guess (Heuristic)
            ans2 = kernel.robust_guess(inp) if χ_score < 1.0 else ans1
            
            task_predictions.append({
                "attempt_1": ans1.tolist(),
                "attempt_2": ans2.tolist()
            })
        submission[task_id] = task_predictions

    # 3. Artifact Generation
    with open('submission.json', 'w') as f:
        json.dump(submission, f)
    print(f"[✓] KERNEL SUCCESS: submission.json created for {len(submission)} tasks.")

if __name__ == "__main__":
    deploy()
```
