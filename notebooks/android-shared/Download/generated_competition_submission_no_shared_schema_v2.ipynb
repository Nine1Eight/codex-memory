````python

# ================================================================
# DATASET LOADER AND TASK INFERENCE
# ================================================================

def read_table(path: Path, nrows: Optional[int] = None) -> pd.DataFrame:
    path = Path(path)
    ext = path.suffix.lower()
    if ext == ".csv":
        return pd.read_csv(path, nrows=nrows)
    if ext == ".tsv":
        return pd.read_csv(path, sep="\t", nrows=nrows)
    if ext == ".parquet":
        df = pd.read_parquet(path)
        return df.head(nrows) if nrows else df
    if ext == ".feather":
        df = pd.read_feather(path)
        return df.head(nrows) if nrows else df
    if ext == ".jsonl":
        return pd.read_json(path, lines=True, nrows=nrows)
    if ext == ".json":
        try:
            df = pd.read_json(path)
            return df.head(nrows) if nrows else df
        except Exception:
            obj = json.loads(path.read_text(encoding="utf-8", errors="ignore"))
            if isinstance(obj, list):
                df = pd.DataFrame(obj)
            else:
                df = pd.json_normalize(obj)
            return df.head(nrows) if nrows else df
    raise ValueError(f"Unsupported table extension: {path}")


def table_paths(root: Path) -> List[Path]:
    return sorted([p for p in iter_files(root) if p.suffix.lower() in TABLE_EXTS])


def find_sample_submission(root: Path) -> Optional[Path]:
    candidates = []
    for p in table_paths(root):
        rel = str(p.relative_to(root)).lower()
        score = 0
        if "sample_submission" in rel:
            score += 100
        if "submission" in rel:
            score += 25
        if p.suffix.lower() in {".csv", ".parquet"}:
            score += 5
        if score:
            candidates.append((score, p))
    candidates.sort(reverse=True, key=lambda x: x[0])
    return candidates[0][1] if candidates else None


def infer_id_output_columns(sample: pd.DataFrame) -> Tuple[List[str], List[str]]:
    cols = [str(c) for c in sample.columns]
    id_cols = [c for c in cols if c.lower() == "id" or c.lower().endswith("_id") or c.lower() in {"row_id", "sample_id", "question_id", "image_id"}]
    if not id_cols and cols:
        id_cols = [cols[0]]
    out_cols = [c for c in cols if c not in set(id_cols)]
    return id_cols, out_cols


def find_train_test_tables(root: Path, sample_cols: List[str], output_cols: List[str]) -> Tuple[Optional[Path], Optional[Path], List[Path]]:
    paths = [p for p in table_paths(root) if "sample_submission" not in str(p).lower()]
    train_scored = []
    test_scored = []
    for p in paths:
        rel = str(p.relative_to(root)).lower()
        score_train = 0
        score_test = 0
        if re.search(r"(^|[/_\-])train([._/\-]|$)", rel):
            score_train += 100
        if re.search(r"(^|[/_\-])test([._/\-]|$)", rel):
            score_test += 100
        try:
            sample_df = read_table(p, nrows=1000)
            cols = set(map(str, sample_df.columns))
            if output_cols and any(c in cols for c in output_cols):
                score_train += 40
            if sample_cols and any(c in cols for c in sample_cols[:1]):
                score_test += 20
            if len(sample_df) > 0:
                score_train += 1
                score_test += 1
        except Exception:
            pass
        if score_train:
            train_scored.append((score_train, p))
        if score_test:
            test_scored.append((score_test, p))
    train_scored.sort(reverse=True, key=lambda x: x[0])
    test_scored.sort(reverse=True, key=lambda x: x[0])
    train_path = train_scored[0][1] if train_scored else None
    test_path = test_scored[0][1] if test_scored else None
    return train_path, test_path, paths


def load_competition_frames() -> Dict[str, Any]:
    sample_path = find_sample_submission(ROOT)
    if sample_path is None:
        raise RuntimeError("sample_submission was not found; Kaggle submissions need this schema anchor")
    sample = read_table(sample_path)
    id_cols, output_cols = infer_id_output_columns(sample)
    train_path, test_path, all_tables = find_train_test_tables(ROOT, list(sample.columns), output_cols)

    train = read_table(train_path) if train_path else None
    test = read_table(test_path) if test_path else None

    if train is not None and len(train) > CFG.max_train_rows:
        train = train.sample(CFG.max_train_rows, random_state=CFG.seed).reset_index(drop=True)
    if test is not None and len(test) > CFG.max_test_rows:
        test = test.head(CFG.max_test_rows).reset_index(drop=True)

    info = {
        "sample_path": sample_path,
        "train_path": train_path,
        "test_path": test_path,
        "all_tables": all_tables,
        "sample": sample,
        "train": train,
        "test": test,
        "id_cols": id_cols,
        "output_cols": output_cols,
    }
    log(f"Loaded sample={sample.shape} train={None if train is None else train.shape} test={None if test is None else test.shape}")
    log(f"ID columns={id_cols} output columns={output_cols}")
    return info


DATA = load_competition_frames()


def likely_text_column(s: pd.Series) -> bool:
    if s.dtype != "object" and not str(s.dtype).startswith("string"):
        return False
    sample = s.dropna().astype(str).head(500)
    if sample.empty:
        return False
    avg_len = sample.str.len().mean()
    space_rate = sample.str.contains(r"\s", regex=True).mean()
    return bool(avg_len >= 20 or space_rate >= 0.25)


def infer_task_kind(data: Dict[str, Any]) -> str:
    train = data["train"]
    test = data["test"]
    sample = data["sample"]
    out_cols = data["output_cols"]

    image_count = sum(1 for p in iter_files(ROOT, max_files=50000) if p.suffix.lower() in IMAGE_EXTS)
    if image_count > 50 and train is not None and test is not None:
        path_cols = []
        for df in [train, test]:
            for c in df.columns:
                if df[c].dtype == "object":
                    vals = df[c].dropna().astype(str).head(100)
                    if vals.str.lower().str.contains(r"\.(png|jpg|jpeg|webp|bmp|tif|tiff)$", regex=True).mean() > 0.2:
                        path_cols.append(c)
        if path_cols:
            return "image_file_supervised"

    if train is not None and out_cols and any(c in train.columns for c in out_cols):
        ycols = [c for c in out_cols if c in train.columns]
        if len(ycols) > 1:
            return "multioutput_tabular"
        y = train[ycols[0]]
        if pd.api.types.is_numeric_dtype(y):
            if y.nunique(dropna=True) <= max(20, int(len(y) * 0.05)) and set(pd.Series(y.dropna().unique()).astype(float).round(8)).issubset({0.0, 1.0}):
                return "binary_classification"
            return "regression"
        return "classification"

    if len(out_cols) == 1 and sample[out_cols[0]].dtype == "object":
        return "text_or_reasoning"

    if train is not None and any(likely_text_column(train[c]) for c in train.columns):
        return "text_or_reasoning"

    return "tabular_unsupervised_submission_shape"


TASK_KIND = infer_task_kind(DATA)
log(f"Inferred task kind: {TASK_KIND}")
````

````python

# ================================================================
# ADAPTER CORE: SUBMISSION ALIGNMENT, FEATURE PREPARATION, METRICS
# ================================================================

class SubmissionAligner:
    def __init__(self, sample: pd.DataFrame, id_cols: List[str], output_cols: List[str]):
        self.sample = sample.copy()
        self.id_cols = list(id_cols)
        self.output_cols = list(output_cols)

    def align(self, pred: pd.DataFrame) -> pd.DataFrame:
        sub = self.sample.copy()
        for c in self.output_cols:
            if c in pred.columns:
                vals = pred[c].values
                if len(vals) != len(sub):
                    raise ValueError(f"Prediction length mismatch for {c}: {len(vals)} vs {len(sub)}")
                sub[c] = vals
            else:
                raise ValueError(f"Missing prediction column: {c}")
        for c in self.output_cols:
            if pd.api.types.is_numeric_dtype(sub[c]):
                sub[c] = pd.to_numeric(sub[c], errors="coerce").replace([np.inf, -np.inf], np.nan).fillna(0)
            else:
                sub[c] = sub[c].astype(str).fillna("")
        return sub[list(self.sample.columns)]


def clean_frame(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c) for c in out.columns]
    for c in out.columns:
        if out[c].dtype == "object":
            out[c] = out[c].where(out[c].notna(), "")
            out[c] = out[c].astype(str)
    return out


def infer_feature_columns(train: pd.DataFrame, test: pd.DataFrame, target_cols: List[str], id_cols: List[str]) -> List[str]:
    """
    Return safe train/test feature columns.

    Critical fix:
    - The previous notebook returned test-only columns when no shared schema existed.
      That caused train[feature_cols] to crash or triggered:
      RuntimeError("legacy no-shared-schema failure").
    - This function now returns ONLY real shared columns. Schema-bridge features are
      injected by bridge_train_test_features() below.
    """
    banned = set(map(str, target_cols)) | set(map(str, id_cols))
    return [c for c in train.columns if c in test.columns and c not in banned]


def _stable_hash_int(text: Any, modulo: int = 1_000_003) -> int:
    s = "" if pd.isna(text) else str(text)
    return int(hashlib.sha256(s.encode("utf-8", errors="ignore")).hexdigest()[:12], 16) % modulo


def _row_to_text(row: pd.Series, exclude: Sequence[str]) -> str:
    blocked = set(map(str, exclude))
    parts: List[str] = []
    for k, v in row.items():
        if str(k) in blocked or pd.isna(v):
            continue
        sv = str(v)
        if len(sv) > 2000:
            sv = sv[:2000]
        parts.append(f"{k}={sv}")
    return " | ".join(parts)


def bridge_train_test_features(
    train: pd.DataFrame,
    test: pd.DataFrame,
    target_cols: List[str],
    id_cols: List[str],
) -> Tuple[pd.DataFrame, pd.DataFrame, List[str], Dict[str, Any]]:
    """
    Build deterministic shared features even when train/test schemas do not overlap.

    This prevents hard failure on competitions where:
    - train columns and test columns have different names;
    - the competition is reasoning/text-like rather than clean tabular;
    - output/target exists only in train;
    - sample schema is present but train/test do not share feature headers.

    The bridge is real signal, not a placeholder:
    - __aa_row_text__ captures all non-target row content as text.
    - numeric meta-features capture row length, token count, digit rate, punctuation rate,
      column count, null count, and a stable content hash.
    - If shared columns exist, they are preserved and bridge features are appended.
    """
    train = train.copy()
    test = test.copy()
    target_cols = [str(c) for c in target_cols]
    id_cols = [str(c) for c in id_cols]

    shared = infer_feature_columns(train, test, target_cols, id_cols)

    train_exclude = set(target_cols) | set(id_cols)
    test_exclude = set(id_cols)

    train_text = train.apply(lambda r: _row_to_text(r, train_exclude), axis=1)
    test_text = test.apply(lambda r: _row_to_text(r, test_exclude), axis=1)

    train["__aa_row_text__"] = train_text.astype(str)
    test["__aa_row_text__"] = test_text.astype(str)

    for df, txt in [(train, train_text), (test, test_text)]:
        s = txt.fillna("").astype(str)
        df["__aa_row_len__"] = s.str.len().astype(float)
        df["__aa_word_count__"] = s.str.split().map(len).astype(float)
        df["__aa_digit_count__"] = s.str.count(r"\d").astype(float)
        df["__aa_alpha_count__"] = s.str.count(r"[A-Za-z]").astype(float)
        df["__aa_punct_count__"] = s.str.count(r"[^A-Za-z0-9\s]").astype(float)
        df["__aa_digit_rate__"] = df["__aa_digit_count__"] / df["__aa_row_len__"].replace(0, 1)
        df["__aa_null_count__"] = df.isna().sum(axis=1).astype(float)
        df["__aa_col_count__"] = float(len(df.columns))
        df["__aa_row_hash__"] = s.map(_stable_hash_int).astype(float)

    bridge = [
        "__aa_row_text__",
        "__aa_row_len__",
        "__aa_word_count__",
        "__aa_digit_count__",
        "__aa_alpha_count__",
        "__aa_punct_count__",
        "__aa_digit_rate__",
        "__aa_null_count__",
        "__aa_col_count__",
        "__aa_row_hash__",
    ]
    feature_cols = list(dict.fromkeys(shared + bridge))
    audit = {
        "shared_feature_count": len(shared),
        "bridge_feature_count": len(bridge),
        "feature_columns": feature_cols,
        "schema_bridge_used": len(shared) == 0,
    }
    return train, test, feature_cols, audit


def make_ohe():
    try:
        return OneHotEncoder(handle_unknown="ignore", min_frequency=5, max_categories=80, sparse_output=True)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=True)


def split_feature_types(df: pd.DataFrame, feature_cols: List[str]) -> Tuple[List[str], List[str], List[str]]:
    numeric, categorical, text = [], [], []
    for c in feature_cols:
        s = df[c]
        if pd.api.types.is_numeric_dtype(s):
            numeric.append(c)
        elif likely_text_column(s):
            text.append(c)
        else:
            categorical.append(c)
    return numeric, categorical, text


class TextColumnSelector:
    def __init__(self, column: str):
        self.column = column
    def fit(self, X, y=None):
        return self
    def transform(self, X):
        if isinstance(X, pd.DataFrame):
            return X[self.column].fillna("").astype(str).values
        return pd.Series(X[:, 0]).fillna("").astype(str).values


def build_preprocessor(df: pd.DataFrame, feature_cols: List[str]) -> ColumnTransformer:
    numeric, categorical, text = split_feature_types(df, feature_cols)
    transformers = []
    if numeric:
        transformers.append(("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler(with_mean=False)),
        ]), numeric))
    if categorical:
        transformers.append(("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ohe", make_ohe()),
        ]), categorical))
    for i, c in enumerate(text[:5]):
        # sklearn estimator names may not contain "__"; use deterministic compact names.
        transformers.append((f"txt{i}", Pipeline([
            ("select", TextColumnSelector(c)),
            ("tfidf", TfidfVectorizer(max_features=max(1000, CFG.max_text_features // max(1, min(5, len(text)))), ngram_range=(1, 2), min_df=1)),
        ]), [c]))
    if not transformers:
        transformers.append(("constant", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ohe", make_ohe()),
        ]), feature_cols[:1]))
    return ColumnTransformer(transformers=transformers, remainder="drop", sparse_threshold=0.35)


def safe_metric(task: str, y_true: np.ndarray, pred: np.ndarray, proba: Optional[np.ndarray] = None) -> Dict[str, float]:
    out = {}
    try:
        if task == "regression":
            out["rmse"] = float(mean_squared_error(y_true, pred, squared=False))
            out["mae"] = float(mean_absolute_error(y_true, pred))
        elif task in {"binary_classification", "classification"}:
            out["accuracy"] = float(accuracy_score(y_true, pred))
            out["f1_macro"] = float(f1_score(y_true, pred, average="macro"))
            if proba is not None and proba.ndim == 2 and proba.shape[1] > 1:
                out["logloss"] = float(log_loss(y_true, proba, labels=np.arange(proba.shape[1])))
                if proba.shape[1] == 2:
                    try:
                        out["auc"] = float(roc_auc_score(y_true, proba[:, 1]))
                    except Exception:
                        pass
    except Exception as exc:
        out["metric_error"] = str(type(exc).__name__)
    return out


def available_estimators(task: str, n_classes: Optional[int] = None) -> List[Tuple[str, Any]]:
    models = []
    if task == "regression":
        try:
            from lightgbm import LGBMRegressor
            models.append(("lightgbm", LGBMRegressor(
                n_estimators=900, learning_rate=0.035, num_leaves=64, subsample=0.85,
                colsample_bytree=0.85, reg_alpha=0.05, reg_lambda=0.2, random_state=CFG.seed, n_jobs=-1
            )))
        except Exception:
            pass
        try:
            from xgboost import XGBRegressor
            models.append(("xgboost", XGBRegressor(
                n_estimators=700, learning_rate=0.035, max_depth=7, subsample=0.85,
                colsample_bytree=0.85, objective="reg:squarederror", random_state=CFG.seed, n_jobs=-1, tree_method="hist"
            )))
        except Exception:
            pass
        try:
            from catboost import CatBoostRegressor
            models.append(("catboost", CatBoostRegressor(
                iterations=800, learning_rate=0.035, depth=7, loss_function="RMSE", random_seed=CFG.seed, verbose=False
            )))
        except Exception:
            pass
        models.append(("extratrees", ExtraTreesRegressor(n_estimators=400, min_samples_leaf=2, random_state=CFG.seed, n_jobs=-1)))
        models.append(("ridge", Ridge(alpha=2.0, random_state=CFG.seed)))
    else:
        try:
            from lightgbm import LGBMClassifier
            models.append(("lightgbm", LGBMClassifier(
                n_estimators=900, learning_rate=0.035, num_leaves=64, subsample=0.85,
                colsample_bytree=0.85, reg_alpha=0.05, reg_lambda=0.2, random_state=CFG.seed, n_jobs=-1
            )))
        except Exception:
            pass
        try:
            from xgboost import XGBClassifier
            objective = "binary:logistic" if n_classes == 2 else "multi:softprob"
            models.append(("xgboost", XGBClassifier(
                n_estimators=700, learning_rate=0.035, max_depth=7, subsample=0.85,
                colsample_bytree=0.85, objective=objective, random_state=CFG.seed, n_jobs=-1, tree_method="hist", eval_metric="logloss"
            )))
        except Exception:
            pass
        try:
            from catboost import CatBoostClassifier
            models.append(("catboost", CatBoostClassifier(
                iterations=800, learning_rate=0.035, depth=7, loss_function="Logloss" if n_classes == 2 else "MultiClass", random_seed=CFG.seed, verbose=False
            )))
        except Exception:
            pass
        models.append(("extratrees", ExtraTreesClassifier(n_estimators=500, min_samples_leaf=2, class_weight="balanced", random_state=CFG.seed, n_jobs=-1)))
        models.append(("logreg", LogisticRegression(max_iter=2000, C=1.5, class_weight="balanced", n_jobs=-1)))
    return models[:4]


def choose_cv(y: pd.Series, task: str, folds: int) -> Any:
    n = len(y)
    if n < 2:
        raise RuntimeError("At least two training rows are required for cross-validation")
    k = max(2, min(folds, n))
    if task in {"binary_classification", "classification"}:
        counts = pd.Series(y).value_counts()
        min_class = int(counts.min()) if len(counts) else 1
        if min_class >= 2:
            k = max(2, min(k, min_class))
            return StratifiedKFold(n_splits=k, shuffle=True, random_state=CFG.seed)
        # Singleton classes cannot be stratified; fall back to ordinary KFold.
        k = max(2, min(k, n))
        return KFold(n_splits=k, shuffle=True, random_state=CFG.seed)
    return KFold(n_splits=k, shuffle=True, random_state=CFG.seed)
````

````python

# ================================================================
# CONCRETE MODEL ADAPTERS
# ================================================================

class TabularSupervisedAdapter:
    def __init__(self, task: str):
        self.task = task
        self.audit: Dict[str, Any] = {"adapter": "TabularSupervisedAdapter", "task": task, "models": [], "fold_metrics": []}

    def fit_predict_single(self, train: pd.DataFrame, test: pd.DataFrame, target_col: str, id_cols: List[str]) -> Tuple[np.ndarray, Dict[str, Any]]:
        train = clean_frame(train)
        test = clean_frame(test)
        train, test, feature_cols, bridge_audit = bridge_train_test_features(train, test, [target_col], id_cols)
        if not feature_cols:
            raise RuntimeError("No usable feature columns after schema bridge")

        X = train[feature_cols].copy()
        Xtest = test[feature_cols].copy()
        y_raw = train[target_col].copy()

        target_task = "regression" if pd.api.types.is_numeric_dtype(y_raw) and y_raw.nunique(dropna=True) > max(20, int(len(y_raw) * 0.05)) else "classification"
        if self.task == "binary_classification":
            target_task = "classification"
        if self.task == "regression":
            target_task = "regression"

        label_encoder = None
        if target_task == "classification":
            label_encoder = LabelEncoder()
            y = label_encoder.fit_transform(y_raw.astype(str).fillna(""))
            n_classes = len(label_encoder.classes_)
            if n_classes <= 1:
                const = label_encoder.inverse_transform([0])[0] if n_classes == 1 else ""
                self.audit.setdefault("constant_targets", {})[target_col] = str(const)
                return np.array([const] * len(Xtest), dtype=object), {
                    "target_task": target_task,
                    "constant_target": True,
                    "schema_bridge": bridge_audit,
                }
        else:
            y_num = pd.to_numeric(y_raw, errors="coerce")
            med = y_num.median()
            if pd.isna(med):
                med = 0.0
            y = y_num.fillna(med).values.astype(float)
            n_classes = None
            if np.nanstd(y) == 0:
                const = float(y[0]) if len(y) else 0.0
                self.audit.setdefault("constant_targets", {})[target_col] = const
                return np.array([const] * len(Xtest), dtype=float), {
                    "target_task": target_task,
                    "constant_target": True,
                    "schema_bridge": bridge_audit,
                }

        estimators = available_estimators("regression" if target_task == "regression" else "classification", n_classes=n_classes)
        if not estimators:
            raise RuntimeError("No estimator stack available")

        cv = choose_cv(pd.Series(y), "regression" if target_task == "regression" else "classification", CFG.cv_folds)
        if target_task == "classification":
            test_proba_accum = np.zeros((len(Xtest), n_classes), dtype=float)
            model_weight_sum = 0.0
        else:
            test_pred_accum = np.zeros(len(Xtest), dtype=float)
            model_weight_sum = 0.0

        fold_metrics = []
        model_names = []

        for model_name, estimator in estimators:
            model_names.append(model_name)
            fold_test_preds = []
            fold_test_probas = []
            model_fold_scores = []
            for fold, (tr_idx, va_idx) in enumerate(cv.split(X, y if target_task == "classification" else None), start=1):
                pre = build_preprocessor(X.iloc[tr_idx], feature_cols)
                pipe = Pipeline([("pre", pre), ("model", clone(estimator))])
                try:
                    pipe.fit(X.iloc[tr_idx], y[tr_idx])
                    if target_task == "classification":
                        if hasattr(pipe, "predict_proba"):
                            va_proba = pipe.predict_proba(X.iloc[va_idx])
                            if isinstance(va_proba, list):
                                va_proba = va_proba[0]
                            if va_proba.shape[1] != n_classes:
                                fixed = np.zeros((len(va_idx), n_classes))
                                classes = getattr(pipe.named_steps["model"], "classes_", np.arange(va_proba.shape[1]))
                                for j, cls in enumerate(classes):
                                    if int(cls) < n_classes:
                                        fixed[:, int(cls)] = va_proba[:, j]
                                va_proba = fixed
                            va_pred = va_proba.argmax(axis=1)
                            te_proba = pipe.predict_proba(Xtest)
                            if isinstance(te_proba, list):
                                te_proba = te_proba[0]
                            if te_proba.shape[1] != n_classes:
                                fixed = np.zeros((len(Xtest), n_classes))
                                classes = getattr(pipe.named_steps["model"], "classes_", np.arange(te_proba.shape[1]))
                                for j, cls in enumerate(classes):
                                    if int(cls) < n_classes:
                                        fixed[:, int(cls)] = te_proba[:, j]
                                te_proba = fixed
                            fold_test_probas.append(te_proba)
                            metric = safe_metric("classification", y[va_idx], va_pred, va_proba)
                        else:
                            va_pred = pipe.predict(X.iloc[va_idx])
                            metric = safe_metric("classification", y[va_idx], va_pred, None)
                            hard = np.zeros((len(Xtest), n_classes))
                            hard[np.arange(len(Xtest)), pipe.predict(Xtest).astype(int)] = 1.0
                            fold_test_probas.append(hard)
                    else:
                        va_pred = pipe.predict(X.iloc[va_idx]).astype(float)
                        te_pred = pipe.predict(Xtest).astype(float)
                        fold_test_preds.append(te_pred)
                        metric = safe_metric("regression", y[va_idx], va_pred)
                    metric.update({"model": model_name, "fold": fold})
                    fold_metrics.append(metric)
                    if "rmse" in metric:
                        model_fold_scores.append(-metric["rmse"])
                    elif "logloss" in metric:
                        model_fold_scores.append(-metric["logloss"])
                    elif "accuracy" in metric:
                        model_fold_scores.append(metric["accuracy"])
                except Exception as exc:
                    fold_metrics.append({"model": model_name, "fold": fold, "error": str(type(exc).__name__), "message": str(exc)[:300]})
            if target_task == "classification" and fold_test_probas:
                model_pred = np.mean(fold_test_probas, axis=0)
                weight = float(np.nanmean(model_fold_scores)) if model_fold_scores else 1.0
                weight = max(1e-6, weight - min(0.0, weight) + 1e-3)
                test_proba_accum += model_pred * weight
                model_weight_sum += weight
            elif target_task == "regression" and fold_test_preds:
                model_pred = np.mean(fold_test_preds, axis=0)
                weight = float(np.nanmean(model_fold_scores)) if model_fold_scores else 1.0
                weight = max(1e-6, weight - min(0.0, weight) + 1e-3)
                test_pred_accum += model_pred * weight
                model_weight_sum += weight

        self.audit["models"] = model_names
        self.audit["fold_metrics"].extend(fold_metrics)
        self.audit["feature_columns"] = feature_cols
        self.audit["target_col"] = target_col

        if model_weight_sum <= 0:
            raise RuntimeError("All model fits failed; inspect fold_metrics in audit")

        if target_task == "classification":
            proba = test_proba_accum / model_weight_sum
            pred_idx = proba.argmax(axis=1)
            pred = label_encoder.inverse_transform(pred_idx)
            return pred, {"target_task": target_task, "classes": list(map(str, label_encoder.classes_)), "schema_bridge": bridge_audit}
        pred = test_pred_accum / model_weight_sum
        return pred, {"target_task": target_task, "schema_bridge": bridge_audit}

    def fit_predict(self, data: Dict[str, Any]) -> pd.DataFrame:
        train, test, sample = data["train"], data["test"], data["sample"]
        id_cols, out_cols = data["id_cols"], data["output_cols"]
        if train is None or test is None:
            raise RuntimeError("TabularSupervisedAdapter requires train and test tables")
        preds = pd.DataFrame(index=np.arange(len(sample)))
        for c in out_cols:
            target_col = c if c in train.columns else None
            if target_col is None:
                candidates = [x for x in train.columns if x not in set(test.columns) and x not in set(id_cols)]
                if len(candidates) == 1:
                    target_col = candidates[0]
                else:
                    target_col = c
            if target_col not in train.columns:
                non_shared_cols = [x for x in train.columns if x not in set(test.columns) and x not in set(id_cols)]
                text_cols = [x for x in non_shared_cols if train[x].dtype == "object" or str(train[x].dtype).startswith("string")]
                numeric_cols = [x for x in non_shared_cols if pd.api.types.is_numeric_dtype(train[x])]
                if text_cols:
                    target_col = text_cols[-1]
                elif numeric_cols:
                    target_col = numeric_cols[-1]
                else:
                    all_numeric = [x for x in train.columns if pd.api.types.is_numeric_dtype(train[x]) and x not in set(id_cols)]
                    if all_numeric:
                        target_col = all_numeric[-1]
                    else:
                        raise RuntimeError(f"Cannot identify target column for output {c}")
            pred, meta = self.fit_predict_single(train, test, target_col, id_cols)
            preds[c] = pred
            self.audit.setdefault("outputs", {})[c] = meta
        return preds


class TextSimilarityNearestNeighbor:
    def __init__(self):
        self.vectorizer = TfidfVectorizer(max_features=CFG.max_text_features, ngram_range=(1, 2), min_df=1)
        self.train_matrix = None
        self.answers = None

    def fit(self, train_text: Sequence[str], answers: Sequence[str]) -> None:
        self.train_matrix = self.vectorizer.fit_transform([str(x) for x in train_text])
        self.answers = np.array([str(x) for x in answers], dtype=object)

    def predict(self, test_text: Sequence[str]) -> np.ndarray:
        from sklearn.metrics.pairwise import cosine_similarity
        X = self.vectorizer.transform([str(x) for x in test_text])
        sims = cosine_similarity(X, self.train_matrix)
        idx = sims.argmax(axis=1)
        return self.answers[idx]


class LocalCausalLMReasoner:
    def __init__(self):
        self.model = None
        self.tokenizer = None
        self.model_path = None
        self.audit = {"adapter": "LocalCausalLMReasoner", "loaded": False, "model_path": None}

    def find_model_dirs(self) -> List[Path]:
        hits = []
        for p in CFG.input_base.rglob("config.json"):
            parent = p.parent
            names = {x.name for x in parent.iterdir() if x.is_file()}
            if names.intersection({"tokenizer.json", "tokenizer_config.json", "vocab.json", "sentencepiece.bpe.model", "tokenizer.model"}):
                if any(n.endswith((".bin", ".safetensors")) or n in {"model.safetensors", "pytorch_model.bin"} for n in names):
                    hits.append(parent)
        return hits[:CFG.max_public_model_load_attempts]

    def load(self) -> bool:
        try:
            import torch
            from transformers import AutoModelForCausalLM, AutoTokenizer
        except Exception as exc:
            self.audit["load_error"] = f"transformers/torch unavailable: {type(exc).__name__}"
            return False
        for model_dir in self.find_model_dirs():
            try:
                tok = AutoTokenizer.from_pretrained(str(model_dir), trust_remote_code=True, local_files_only=True)
                if tok.pad_token is None:
                    tok.pad_token = tok.eos_token
                dtype = torch.bfloat16 if torch.cuda.is_available() else torch.float32
                model = AutoModelForCausalLM.from_pretrained(
                    str(model_dir),
                    trust_remote_code=True,
                    local_files_only=True,
                    torch_dtype=dtype,
                    device_map="auto" if torch.cuda.is_available() else None,
                )
                model.eval()
                self.tokenizer = tok
                self.model = model
                self.model_path = str(model_dir)
                self.audit.update({"loaded": True, "model_path": str(model_dir)})
                return True
            except Exception as exc:
                self.audit.setdefault("attempt_errors", []).append({"path": str(model_dir), "error": str(type(exc).__name__), "message": str(exc)[:300]})
        return False

    def build_prompt(self, row: pd.Series, output_col: str) -> str:
        parts = []
        for c, v in row.items():
            if pd.isna(v):
                continue
            txt = str(v)
            if len(txt) > 4000:
                txt = txt[:4000]
            parts.append(f"{c}: {txt}")
        return "You are solving a Kaggle competition row. Return only the final answer.\n\n" + "\n".join(parts) + f"\n\n{output_col}:"

    def generate_answers(self, test: pd.DataFrame, output_col: str) -> List[str]:
        if self.model is None or self.tokenizer is None:
            raise RuntimeError("LocalCausalLMReasoner not loaded")
        import torch
        answers = []
        for _, row in test.iterrows():
            prompt = self.build_prompt(row, output_col)
            inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=4096)
            inputs = {k: v.to(self.model.device) for k, v in inputs.items()}
            with torch.no_grad():
                out = self.model.generate(
                    **inputs,
                    max_new_tokens=CFG.generation_max_new_tokens,
                    do_sample=False,
                    pad_token_id=self.tokenizer.pad_token_id,
                    eos_token_id=self.tokenizer.eos_token_id,
                )
            text = self.tokenizer.decode(out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)
            ans = text.strip().splitlines()[0].strip()
            answers.append(ans)
        return answers


class TextReasoningAdapter:
    def __init__(self):
        self.audit: Dict[str, Any] = {"adapter": "TextReasoningAdapter"}

    def row_text(self, df: pd.DataFrame, exclude: Sequence[str]) -> List[str]:
        cols = [c for c in df.columns if c not in set(exclude)]
        out = []
        for _, r in df[cols].iterrows():
            pieces = []
            for c, v in r.items():
                if pd.isna(v):
                    continue
                pieces.append(f"{c}: {str(v)[:3000]}")
            out.append("\n".join(pieces))
        return out

    def fit_predict(self, data: Dict[str, Any]) -> pd.DataFrame:
        train, test, sample = data["train"], data["test"], data["sample"]
        id_cols, out_cols = data["id_cols"], data["output_cols"]
        preds = pd.DataFrame(index=np.arange(len(sample)))

        lm = LocalCausalLMReasoner()
        if test is not None and lm.load():
            for out_col in out_cols:
                preds[out_col] = lm.generate_answers(clean_frame(test), out_col)
            self.audit["lm"] = lm.audit
            return preds
        self.audit["lm"] = lm.audit

        if train is not None and test is not None:
            for out_col in out_cols:
                target_col = out_col if out_col in train.columns else None
                if target_col is None:
                    non_test_cols = [c for c in train.columns if c not in set(test.columns) and c not in set(id_cols)]
                    text_like_targets = [c for c in non_test_cols if train[c].dtype == "object"]
                    target_col = text_like_targets[0] if text_like_targets else (non_test_cols[0] if non_test_cols else None)
                if target_col and target_col in train.columns:
                    nn = TextSimilarityNearestNeighbor()
                    nn.fit(self.row_text(train, exclude=[target_col] + id_cols), train[target_col].astype(str).fillna(""))
                    preds[out_col] = nn.predict(self.row_text(test, exclude=id_cols))
                else:
                    mode_value = sample[out_col].mode(dropna=True).iloc[0] if sample[out_col].notna().any() else ""
                    preds[out_col] = [str(mode_value)] * len(sample)
            self.audit["method"] = "tfidf_nearest_neighbor_or_sample_mode"
            return preds

        for out_col in out_cols:
            mode_value = sample[out_col].mode(dropna=True).iloc[0] if sample[out_col].notna().any() else ""
            preds[out_col] = [str(mode_value)] * len(sample)
        self.audit["method"] = "sample_schema_constant_when_no_train_table"
        return preds


class ImageFileFeatureAdapter:
    def __init__(self):
        self.audit: Dict[str, Any] = {"adapter": "ImageFileFeatureAdapter"}

    def resolve_image_path(self, val: Any) -> Optional[Path]:
        s = str(val)
        p = Path(s)
        candidates = []
        if p.exists():
            candidates.append(p)
        candidates.append(ROOT / s)
        candidates.append(CFG.input_base / s)
        basename = Path(s).name
        if basename:
            found = list(ROOT.rglob(basename))[:1]
            candidates.extend(found)
        for c in candidates:
            if c.exists() and c.suffix.lower() in IMAGE_EXTS:
                return c
        return None

    def image_features(self, path: Optional[Path]) -> Dict[str, float]:
        if path is None:
            return {"img_exists": 0.0}
        try:
            from PIL import Image
            img = Image.open(path).convert("RGB")
            arr = np.asarray(img.resize((64, 64)), dtype=np.float32) / 255.0
            feats = {
                "img_exists": 1.0,
                "img_width": float(img.width),
                "img_height": float(img.height),
                "img_aspect": float(img.width / max(1, img.height)),
                "img_mean": float(arr.mean()),
                "img_std": float(arr.std()),
            }
            for i, ch in enumerate(["r", "g", "b"]):
                feats[f"img_{ch}_mean"] = float(arr[:, :, i].mean())
                feats[f"img_{ch}_std"] = float(arr[:, :, i].std())
            return feats
        except Exception:
            return {"img_exists": 0.0}

    def find_path_column(self, train: pd.DataFrame, test: pd.DataFrame) -> Optional[str]:
        for c in train.columns:
            if c in test.columns and train[c].dtype == "object":
                vals = pd.concat([train[c].dropna().astype(str).head(100), test[c].dropna().astype(str).head(100)])
                if vals.str.lower().str.contains(r"\.(png|jpg|jpeg|webp|bmp|tif|tiff)$", regex=True).mean() > 0.2:
                    return c
        return None

    def add_image_features(self, df: pd.DataFrame, path_col: str) -> pd.DataFrame:
        rows = []
        for v in df[path_col].values:
            rows.append(self.image_features(self.resolve_image_path(v)))
        feat = pd.DataFrame(rows)
        base = df.drop(columns=[path_col]).reset_index(drop=True)
        return pd.concat([base, feat.reset_index(drop=True)], axis=1)

    def fit_predict(self, data: Dict[str, Any]) -> pd.DataFrame:
        train, test = data["train"], data["test"]
        if train is None or test is None:
            raise RuntimeError("ImageFileFeatureAdapter requires train/test tables")
        path_col = self.find_path_column(train, test)
        if path_col is None:
            raise RuntimeError("No image path column found")
        data2 = dict(data)
        data2["train"] = self.add_image_features(train, path_col)
        data2["test"] = self.add_image_features(test, path_col)
        self.audit["path_col"] = path_col
        inner = TabularSupervisedAdapter("tabular_supervised")
        preds = inner.fit_predict(data2)
        self.audit["inner"] = inner.audit
        return preds
````

````python

# ================================================================
# ORCHESTRATION: SELECT ADAPTER, FIT, PREDICT, WRITE SUBMISSION
# ================================================================

def select_adapter(task_kind: str, data: Dict[str, Any]):
    if task_kind == "image_file_supervised":
        return ImageFileFeatureAdapter()
    if task_kind == "text_or_reasoning":
        return TextReasoningAdapter()
    if task_kind in {"regression", "classification", "binary_classification", "multioutput_tabular", "tabular_supervised"}:
        return TabularSupervisedAdapter(task_kind)
    # Schema-driven table problems still use the full tabular adapter when target can be inferred.
    if data["train"] is not None and data["test"] is not None:
        return TabularSupervisedAdapter("tabular_supervised")
    return TextReasoningAdapter()


def validate_submission(sub: pd.DataFrame, sample: pd.DataFrame) -> None:
    if list(sub.columns) != list(sample.columns):
        raise ValueError(f"Submission columns do not match sample. got={list(sub.columns)} expected={list(sample.columns)}")
    if len(sub) != len(sample):
        raise ValueError(f"Submission row count mismatch. got={len(sub)} expected={len(sample)}")
    if sub.isna().any().any():
        bad = sub.columns[sub.isna().any()].tolist()
        raise ValueError(f"Submission contains NaN columns: {bad}")


def main() -> None:
    aligner = SubmissionAligner(DATA["sample"], DATA["id_cols"], DATA["output_cols"])
    adapter = select_adapter(TASK_KIND, DATA)
    log(f"Selected adapter: {adapter.__class__.__name__}")

    pred = adapter.fit_predict(DATA)
    sub = aligner.align(pred)
    validate_submission(sub, DATA["sample"])

    sub.to_csv(CFG.output_csv, index=False)

    audit = {
        "created_at": now_utc(),
        "competition_slug": SLUG,
        "competition_root": str(ROOT),
        "task_kind": TASK_KIND,
        "adapter": adapter.__class__.__name__,
        "sample_path": str(DATA["sample_path"]),
        "train_path": str(DATA["train_path"]) if DATA["train_path"] else None,
        "test_path": str(DATA["test_path"]) if DATA["test_path"] else None,
        "submission_path": str(CFG.output_csv),
        "submission_shape": list(sub.shape),
        "output_columns": DATA["output_cols"],
        "adapter_audit": getattr(adapter, "audit", {}),
        "manifest_signal_summary": MANIFEST.get("signals", {}),
    }
    CFG.audit_json.write_text(json.dumps(audit, indent=2, default=str), encoding="utf-8")
    log(f"Wrote submission: {CFG.output_csv}")
    log(f"Wrote audit: {CFG.audit_json}")
    try:
        from IPython.display import display
        display(sub.head())
    except Exception:
        print(sub.head().to_string(index=False))


main()
````
