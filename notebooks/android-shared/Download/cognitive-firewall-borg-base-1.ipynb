Hardened base notebook for the JED attack-algorithm server contract. It preserves the uploaded attack template and removes tabular `sample_submission.csv` assumptions.


```python
import glob
import os
import sys
from pathlib import Path

# --- Cognitive Firewall / JED Attack notebook bootstrap ---
# No sample_submission.csv dependency. This notebook is for the agent-security
# attack-algorithm server contract, not a tabular Kaggle CSV competition.
sys.argv = [sys.argv[0]]
PROJECT_ROOT = Path.cwd()


def _add_path_once(path: Path | str) -> None:
    s = str(path)
    if s and s not in sys.path:
        sys.path.insert(0, s)


def _find_sdk_root() -> str | None:
    # Fast exact locations first.
    roots = [
        Path('/kaggle/input'),
        PROJECT_ROOT / 'data',
        PROJECT_ROOT,
        Path.cwd() / 'data',
        Path('/mnt/data'),
    ]
    for root in roots:
        if not root.exists():
            continue
        if (root / 'aicomp_sdk').exists() and (root / 'kaggle_evaluation').exists():
            return str(root)

    # Kaggle datasets often mount nested SDK folders.
    for pattern in (
        '/kaggle/input/**/kaggle_evaluation',
        '/kaggle/input/**/aicomp_sdk',
        '/mnt/data/**/kaggle_evaluation',
    ):
        for candidate in glob.glob(pattern, recursive=True):
            candidate = Path(candidate)
            parent = candidate.parent
            if (parent / 'aicomp_sdk').exists() and (parent / 'kaggle_evaluation').exists():
                return str(parent)
    return None


dataset_root = _find_sdk_root()
if dataset_root is not None:
    _add_path_once(dataset_root)

try:
    import aicomp_sdk  # noqa: F401
    import kaggle_evaluation  # noqa: F401
    SDK_AVAILABLE = True
except Exception as _sdk_error:
    SDK_AVAILABLE = False
    SDK_ERROR = repr(_sdk_error)
else:
    SDK_ERROR = ''

WORKING_DIR = Path('/kaggle/working') if Path('/kaggle/working').exists() else PROJECT_ROOT
WORKING_DIR.mkdir(parents=True, exist_ok=True)
ATTACK_PATH = WORKING_DIR / 'attack.py'
SUBMISSION_PATH = WORKING_DIR / 'submission.csv'
STATUS_PATH = WORKING_DIR / 'cognitive_firewall_status.json'

print(f'PROJECT_ROOT : {PROJECT_ROOT}')
print(f'WORKING_DIR  : {WORKING_DIR}')
print(f'SDK root     : {dataset_root or "<not found in public run>"}')
print(f'SDK available: {SDK_AVAILABLE}')
if not SDK_AVAILABLE:
    print(f'SDK error    : {SDK_ERROR}')
```

```python
import importlib.util
import json
import py_compile
import time

attack_code = ATTACK_TEMPLATE.replace('__SEARCH_PROFILE__', repr(SEARCH_PROFILE))
ATTACK_PATH.write_text(attack_code + '\n', encoding='utf-8')
py_compile.compile(str(ATTACK_PATH), doraise=True)

status = {
    'profile': SEARCH_PROFILE,
    'attack_path': str(ATTACK_PATH),
    'submission_path': str(SUBMISSION_PATH),
    'sdk_root': dataset_root,
    'sdk_available': bool(SDK_AVAILABLE),
    'sdk_error': SDK_ERROR,
    'generated_at_unix': time.time(),
    'contract_checked': False,
}

if SDK_AVAILABLE:
    spec = importlib.util.spec_from_file_location('attack_contract_check', ATTACK_PATH)
    attack = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(attack)

    class DummyConfig:
        time_budget_s = 1.0
        max_steps = 0
        max_tool_hops = 4

    algorithm = attack.AttackAlgorithm()
    static_candidates = algorithm.run(None, DummyConfig())

    # contract assertions
    assert attack.SEARCH_PROFILE == SEARCH_PROFILE
    assert attack.RETURN_TARGET_CANDIDATES >= 60
    assert len(static_candidates) == attack.STATIC_FALLBACK_CANDIDATES
    assert attack.MAX_VERIFIED_CANDIDATES >= 1
    assert attack.ENDPOINTS_PER_MSG >= 1
    assert len(static_candidates) > 0

    import re as _re
    _urls = [u for c in static_candidates for m in c.user_messages
             for u in _re.findall(r'https?://[^\s,;]+', m)]
    _regdom = {'.'.join(_re.sub(r'^https?://', '', u).split('/')[0].split('.')[-2:]) for u in _urls}
    _emails = [e for c in static_candidates for m in c.user_messages
               for e in _re.findall(r'[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}', m)]
    _email_targets = set(_emails)

    status.update({
        'contract_checked': True,
        'return_target_candidates': int(attack.RETURN_TARGET_CANDIDATES),
        'max_verified_candidates': int(attack.MAX_VERIFIED_CANDIDATES),
        'static_fallback_candidates': int(len(static_candidates)),
        'endpoints_per_msg': int(attack.ENDPOINTS_PER_MSG),
        'max_search_s': float(attack.MAX_SEARCH_S),
        'default_max_hops': int(attack.DEFAULT_MAX_HOPS),
        'max_msg_chars': int(attack.MAX_MSG_CHARS),
        'score_ceiling_estimate': float(attack.RETURN_TARGET_CANDIDATES * 18 / 200),
        'baseline_delta_estimate': float((attack.RETURN_TARGET_CANDIDATES - 620) * 18 / 200),
        'distinct_regdom': int(len(_regdom)),
        'email_targets': int(len(_email_targets)),
        'target_estimate': int(len(_regdom) + len(_email_targets)),
        'sample_msg': static_candidates[0].user_messages[0][:120],
    })

    print(f'profile        : {attack.SEARCH_PROFILE}')
    print(f'return_target  : {attack.RETURN_TARGET_CANDIDATES}')
    print(f'max_verified   : {attack.MAX_VERIFIED_CANDIDATES}')
    print(f'endpoints/msg  : {attack.ENDPOINTS_PER_MSG}')
    print(f'max_search_s   : {attack.MAX_SEARCH_S}')
    print(f'max_hops       : {attack.DEFAULT_MAX_HOPS}')
    print(f'max_msg_chars  : {attack.MAX_MSG_CHARS}')
    print(f'score_ceiling  : {attack.RETURN_TARGET_CANDIDATES * 18 / 200:.1f}')
    print(f'baseline_delta : {(attack.RETURN_TARGET_CANDIDATES - 620) * 18 / 200:+.1f} vs 55.8')
    print(f'static fallback: {len(static_candidates)} candidates')
    print(f'distinct regdom: {len(_regdom)}')
    print(f'email targets  : {len(_email_targets)}')
    print(f'target estimate: {len(_regdom) + len(_email_targets)}')
    print(f'sample msg     : {static_candidates[0].user_messages[0][:80]}')
else:
    # Public notebook sessions can lack the hidden SDK mount. This is not a fatal
    # condition for authoring; the hidden rerun must provide the SDK for serving.
    print('SDK unavailable in this environment; attack.py was still generated and py_compile passed.')
    print('Hidden competition rerun will import the SDK and start the inference server.')

STATUS_PATH.write_text(json.dumps(status, indent=2, sort_keys=True) + '\n', encoding='utf-8')
print(f'status written : {STATUS_PATH}')
```
