```python
# =========================
# 1. Install
# =========================
!uv pip install -q --no-cache-dir --force-reinstall \
"tinker-cookbook @ git+https://github.com/thinking-machines-lab/tinker-cookbook.git@nightly"

# =========================
# 2. Imports
# =========================
import torch
import shutil
import os
import gc
import time
import tinker_cookbook.weights._adapter as A
from tinker_cookbook import weights
from dataclasses import dataclass
from typing import Optional

print("✅ Imports done")

# =========================
# 3. Config (upgradeable)
# =========================
@dataclass
class LoRAConfig:
    max_rank: int = 32
    min_rank: int = 4
    device: str = "cpu"
    energy_threshold: float = 0.99     # retain 99% of signal energy
    adaptive_rank: bool = True          # auto-select rank by energy
    stabilize_eps: float = 1e-8
    verbose: bool = True

CFG = LoRAConfig()

# =========================
# 4. Rank selection utility
# =========================
def _select_rank_by_energy(S: torch.Tensor, threshold: float, max_rank: int, min_rank: int) -> int:
    """Choose smallest rank that retains `threshold` fraction of total energy."""
    energy = (S ** 2).cumsum(0) / (S ** 2).sum()
    candidates = (energy >= threshold).nonzero(as_tuple=True)[0]
    if len(candidates) == 0:
        rank = max_rank
    else:
        rank = int(candidates[0].item()) + 1  # 1-indexed
    return max(min_rank, min(rank, max_rank))

# =========================
# 5. Upgraded SVD compression
# =========================
def _compress_lora_pair_to_rank(
    B: torch.Tensor,
    A_mat: torch.Tensor,
    cfg: LoRAConfig = CFG,
) -> tuple[torch.Tensor, torch.Tensor, int]:
    """
    Compress (B @ A) via truncated SVD.
    Returns (B_new, A_new, final_rank).
    """
    t0 = time.time()
    raw_rank = B.shape[1]

    # Compute delta on CPU in float32 for stability
    delta = (B.float() @ A_mat.float()).to(cfg.device)

    # SVD with fallback
    try:
        U, S, Vh = torch.linalg.svd(delta, full_matrices=False)
    except RuntimeError as e:
        if cfg.verbose:
            print(f"  ⚠️  linalg.svd failed ({e}), falling back to torch.svd")
        U, S, Vh = torch.svd(delta)

    # Adaptive or fixed rank
    if cfg.adaptive_rank:
        rank = _select_rank_by_energy(S, cfg.energy_threshold, cfg.max_rank, cfg.min_rank)
    else:
        rank = cfg.max_rank

    # Truncate
    U  = U[:, :rank]
    S  = S[:rank]
    Vh = Vh[:rank, :]

    # Stabilize singular values
    S = torch.clamp(S, min=cfg.stabilize_eps)
    sroot = S.sqrt()

    B_new = (U * sroot.unsqueeze(0)).contiguous()
    A_new = (sroot.unsqueeze(1) * Vh).contiguous()

    # Energy retained (informational)
    energy_kept = float((S ** 2).sum() / (torch.linalg.svd(delta, full_matrices=False)[1] ** 2).sum())

    elapsed = time.time() - t0
    if cfg.verbose:
        print(f"  🔧 {raw_rank} → {rank}  |  energy={energy_kept:.4f}  |  {elapsed:.2f}s")

    # Free memory
    del delta, U, S, Vh
    gc.collect()

    return B_new, A_new, rank

# =========================
# 6. Upgraded merge patch
# =========================
def patched_merge_fused_projections(
    fused_model_key,
    adapter_layer_prefix,
    components,
    model_state_shapes,
    peft_weights,
    target_modules,
    profile,
):
    fused_out_dim = model_state_shapes[fused_model_key][0]
    fused_target_name = fused_model_key.removesuffix(".weight").rsplit(".", 1)[-1]

    # Find mapping
    component_order = None
    for target, comps in profile.fused_projection_map:
        if target == fused_target_name:
            component_order = comps
            break
    if component_order is None:
        raise RuntimeError(f"❌ Missing fused mapping for: {fused_target_name}")

    comp_by_name = {n: (a, b) for n, a, b in components}

    # Validate all components exist upfront
    missing = [n for n in component_order if n not in comp_by_name]
    if missing:
        raise RuntimeError(f"❌ Missing components: {missing}")

    lora_A_parts, comp_slices = [], []
    merged_rank, row_offset = 0, 0

    for name in component_order:
        A_mat, B = comp_by_name[name]
        r       = A_mat.shape[0]
        out_dim = B.shape[0]
        lora_A_parts.append(A_mat)
        comp_slices.append((row_offset, row_offset + out_dim, r, name))
        row_offset  += out_dim
        merged_rank += r

    merged_lora_A = torch.cat(lora_A_parts, dim=0)
    merged_lora_B = torch.zeros(fused_out_dim, merged_rank, dtype=merged_lora_A.dtype)

    rank_offset = 0
    for row_start, row_end, r, name in comp_slices:
        _, B = comp_by_name[name]
        merged_lora_B[row_start:row_end, rank_offset:rank_offset + r] = B
        rank_offset += r

    if CFG.verbose:
        print(f"  📊 Merged rank before compression: {merged_rank}")

    # Compress only if needed
    if merged_rank > CFG.max_rank:
        merged_lora_B, merged_lora_A, final_rank = _compress_lora_pair_to_rank(
            merged_lora_B, merged_lora_A, CFG
        )
    else:
        final_rank = merged_rank
        if CFG.verbose:
            print(f"  ✅ No compression needed (rank={final_rank})")

    peft_key = f"{adapter_layer_prefix}.{fused_target_name}.weight"
    A._add_peft_weight(peft_key, merged_lora_A, merged_lora_B, peft_weights, target_modules)

    if CFG.verbose:
        print(f"  ✅ Final rank: {final_rank}")

    return final_rank

# =========================
# 7. Apply patch
# =========================
A._merge_fused_projections = patched_merge_fused_projections
print("✅ Patch applied")

# =========================
# 8. Build adapter (with timing)
# =========================
OUTPUT_DIR = "/kaggle/working/nemotron-adapter-ready-to-submit"

print("\n🚀 Building adapter...")
t_start = time.time()

weights.build_lora_adapter(
    base_model="nvidia/NVIDIA-Nemotron-3-Nano-30B-A3B-BF16",
    adapter_path="/kaggle/input/models/huikang/nemotron-adapter/transformers/default/20",
    output_path=OUTPUT_DIR,
)

print(f"\n✅ Adapter built in {time.time() - t_start:.1f}s")

# =========================
# 9. Zip submission
# =========================
ZIP_PATH = "/kaggle/working/submission.zip"

if os.path.exists(OUTPUT_DIR):
    print("📦 Zipping...")
    shutil.make_archive(ZIP_PATH.replace(".zip", ""), "zip", OUTPUT_DIR)
    size_mb = os.path.getsize(ZIP_PATH) / 1e6
    print(f"✅ Zipped: {ZIP_PATH}  ({size_mb:.1f} MB)")
else:
    raise FileNotFoundError(f"❌ Output folder missing: {OUTPUT_DIR}")
```
